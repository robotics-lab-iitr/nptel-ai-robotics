## Problem Statement

The initial path point is $[x_1\ y_1\ \phi_1] = [0.700\ \ 0.200\ \ 0^{\circ}]$ and the final path point is $[x_2\ y_2\ \phi_2] = [0.300\ \ {-0.400}\ \ 60^{\circ}]$.

The duration of the segment is $t_f = 4.0\ s$. Assuming **linear interpolation in Cartesian space** with a path-update interval of $\Delta t = 0.2\ s$, we will write a Python program to:

- Generate the Cartesian trajectory
- Compute Cartesian position components: $x(t)$, $y(t)$, $\phi(t)$
- Compute Cartesian velocity components: $\dot{x}(t)$, $\dot{y}(t)$, $\dot{\phi}(t)$
- Compute Cartesian acceleration components: $\ddot{x}(t)$, $\ddot{y}(t)$, $\ddot{\phi}(t)$
- Print the trajectory values for the complete motion interval

## Inputs

Define the segment duration, path-update interval, and the initial/final Cartesian path points.

```python
# Segment duration
tf = 4.0

# Path-update interval
dt = 0.2

# Initial Cartesian point
x1 = 0.700
y1 = 0.200
phi1 = 0.0

# Final Cartesian point
x2 = 0.300
y2 = -0.400
phi2 = 60.0

# Time vector
t = np.arange(0, tf + dt, dt)
```

## Cartesian Position, Velocity, and Acceleration

With **linear interpolation**, each Cartesian coordinate varies linearly with time between its initial and final value. That makes velocity constant over the segment and acceleration zero throughout (aside from the instantaneous jumps at the very start/end, which this simple scheme does not model).

```python
# Cartesian position
x = x1 + ((x2 - x1) / tf) * t
y = y1 + ((y2 - y1) / tf) * t
phi = phi1 + ((phi2 - phi1) / tf) * t

# Cartesian velocity
xdot = np.ones_like(t) * ((x2 - x1) / tf)
ydot = np.ones_like(t) * ((y2 - y1) / tf)
phidot = np.ones_like(t) * ((phi2 - phi1) / tf)

# Cartesian acceleration
xddot = np.zeros_like(t)
yddot = np.zeros_like(t)
phiddot = np.zeros_like(t)
```

## Print the Trajectory Table

Print a formatted table of every quantity at each time step over the full motion interval.

```python
# Print heading
print("-" * 90)
print(" t       x       y       phi     xdot     ydot    phidot   xddot yddot phiddot")
print("-" * 90)

# Print trajectory values
for i in range(len(t)):
    print(f"{t[i]:.1f}   "
          f"{x[i]:.3f}   "
          f"{y[i]:.3f}   "
          f"{phi[i]:.1f}   "
          f"{xdot[i]:.3f}   "
          f"{ydot[i]:.3f}   "
          f"{phidot[i]:.1f}   "
          f"{xddot[i]:.1f}   "
          f"{yddot[i]:.1f}   "
          f"{phiddot[i]:.1f}")
```

## Visualize the Trajectory

Not shown as code on the slides, but a quick plot makes the linear interpolation easy to see: straight-line position curves, constant velocity, and zero acceleration.

```python
fig, axes = plt.subplots(3, 3, figsize=(12, 8), sharex=True)

labels = [
    ('x(t)', x), ('y(t)', y), (r'$\phi(t)$', phi),
    (r'$\dot{x}(t)$', xdot), (r'$\dot{y}(t)$', ydot), (r'$\dot{\phi}(t)$', phidot),
    (r'$\ddot{x}(t)$', xddot), (r'$\ddot{y}(t)$', yddot), (r'$\ddot{\phi}(t)$', phiddot),
]

for ax, (label, data) in zip(axes.flat, labels):
    ax.plot(t, data, marker='o')
    ax.set_title(label)
    ax.grid(True)

for ax in axes[-1]:
    ax.set_xlabel('Time (s)')

plt.tight_layout()
plt.show()
```
