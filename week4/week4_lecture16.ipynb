# Joint-Space Trajectory Generation

This notebook follows along with the cubic polynomial joint-space trajectory example from the slides of Lecture 16: Joint-Space Schemes

```python
try:
    import google.colab
    print('Running on CoLab')
    COLAB = True
except ModuleNotFoundError:
    COLAB = False

import numpy as np
import matplotlib.pyplot as plt
```

We are given boundary conditions on joint angle and joint velocity at the start and end of the motion, and we solve for the coefficients of a cubic polynomial $\theta(t) = a_0 + a_1 t + a_2 t^2 + a_3 t^3$ that satisfies them. We then plot the resulting position, velocity, and acceleration profiles.

## Inputs

Set the boundary conditions: initial and final joint angle, initial and final joint velocity, and the total travel time.

```python
# Inputs
theta_0 = 10
theta_f = 70
theta_dot_0 = 0
theta_dot_f = 0
tf = 5
```

## Coefficients

From the boundary conditions at $t = 0$:

$a_0 = \theta_0$, $a_1 = \dot\theta_0$

The remaining two coefficients ($a_2$, $a_3$) are found by solving a 2x2 linear system built from the conditions at $t = t_f$.

```python
# Coefficients
a0 = theta_0
a1 = theta_dot_0

A = np.array([[tf**2, tf**3],
              [2*tf,  3*tf**2]])

B = np.array([theta_f - a0 - a1*tf,
              theta_dot_f - a1])

[a2, a3] = np.linalg.solve(A, B)
```

## Time and Trajectory

Build the time vector and evaluate the position, velocity, and acceleration polynomials over it.

```python
# Time
t = np.linspace(0, tf, 100)

# Trajectory
theta = a0 + a1*t + a2*t**2 + a3*t**3
theta_dot = a1 + 2*a2*t + 3*a3*t**2
theta_ddot = 2*a2 + 6*a3*t
```

## Plot

Plot the joint position, velocity, and acceleration profiles as three stacked subplots.

```python
# Plot
plt.figure(figsize=(10,6))

# Position
plt.subplot(3,1,1)
plt.plot(t, theta)
plt.title(r'Joint Trajectory $\theta(t)$')
plt.ylabel(r'$\theta(t)$')

# Velocity
plt.subplot(3,1,2)
plt.plot(t, theta_dot)
plt.title(r'Joint Velocity $\dot{\theta}(t)$')
plt.ylabel(r'$\dot{\theta}(t)$')

# Acceleration
plt.subplot(3,1,3)
plt.plot(t, theta_ddot)
plt.title(r'Joint Acceleration $\ddot{\theta}(t)$')
plt.ylabel(r'$\ddot{\theta}(t)$')
plt.xlabel('Time (s)')

plt.tight_layout()
plt.show()
```
