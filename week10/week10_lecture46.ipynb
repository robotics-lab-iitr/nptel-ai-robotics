## Text Pre-processing: In a single go

````python
import nltk
import string

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download required resources (run once)
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt_tab')

# Input command
text = "Pick up the red ball near the chair!"

print("Original Text:")
print(text)

# Step 1: Convert to lowercase
text = text.lower()

print("\nAfter Lowercasing:")
print(text)

# Step 2: Remove punctuation
text = text.translate(str.maketrans('', '', string.punctuation))

print("\nAfter Removing Punctuation:")
print(text)

# Step 3: Tokenization
tokens = word_tokenize(text)

print("\nTokens:")
print(tokens)

# Step 4: Remove stop words
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens
    if word not in stop_words
]

print("\nAfter Stop-word Removal:")
print(filtered_tokens)

# Step 5: Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [
    lemmatizer.lemmatize(word)
    for word in filtered_tokens
]

print("\nAfter Lemmatization:")
print(lemmatized_tokens)
````

## Text Pre-processng: Step-by step explanation

### Lowercasing -> Punctuation Removal -> Tokenization

````python
import nltk
import string

# The Punkt tokenizer is an unsupervised algorithm that learns to tokenize text by identifying sentence boundaries
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

text = "Pick up the red ball near the chair!"
text = text.lower() # convert to lowercase
# remove punctuation
text = text.translate(str.maketrans('', '', string.punctuation))
tokens = word_tokenize(text)
print("tokens:",tokens)
print(type(word_tokenize(text)))
````

### Removing stopwords

````python
from nltk.corpus import stopwords
nltk.download('stopwords')

# Filtering out stopwords
stop_words = set(stopwords.words('english'))
# print("stopwords:",stop_words) # You can see the stopwords
# print(len(stop_words)) # Count the no. of stopwords

filtered_tokens = [w for w in tokens if not w in stop_words]
print(filtered_tokens)
````

### Reduce to root/base form using Lemmatization

````python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [
    lemmatizer.lemmatize(word)
    for word in filtered_tokens
]
print(lemmatized_tokens)
````

#### Why Lemmatization is better than Stemming

````python
from nltk.stem import PorterStemmer, WordNetLemmatizer
nltk.download('wordnet') # wordnet is a corpus

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

word = "studies"
print(f"Stemmed: {stemmer.stem(word)}")
print(f"Lemmatized: {lemmatizer.lemmatize(word)}")
````

## Vectorization

### One-hot encoding

````python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Convert tokens into column format
words = [[word] for word in lemmatized_tokens]

print("\nInput words for One-Hot Encoding:")
print(words)

# One-Hot Encoder
encoder = OneHotEncoder(sparse_output=False)

# Transform
one_hot_vectors = encoder.fit_transform(words)

# Display
df = pd.DataFrame(
    one_hot_vectors,
    columns=encoder.get_feature_names_out()
)

print("\nOne-Hot Encoding Output (DataFrame):")
print(df)
````

### Bag Of Words (BOW)

````python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

commands = [
    "pick red ball near chair",
    "pick blue box",
    "grab red box"
]

# Create BoW vectorizer
vectorizer = CountVectorizer()

# Convert text into BoW vectors
X = vectorizer.fit_transform(commands)

# Display matrix
df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print(df)
````

### TF-IDF

````python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# Commands
commands = [
    "pick red ball near chair",
    "pick blue box",
    "grab red box"
]

# Create TF-IDF Vectorizer
vectorizer = TfidfVectorizer()

# Convert commands into TF-IDF vectors
X = vectorizer.fit_transform(commands)

# Display matrix
df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)

print(df.round(2))
````

## Language Understanding

### POS Tagging for action-object extraction



````python
nltk.download('averaged_perceptron_tagger_eng') # Changed to download the specific 'eng' resource

# Input sentence
command = "Pick up the red ball near the chair"

# Lowercasing + Tokenization
tokens = word_tokenize(command.lower())

# POS Tagging
pos_tags = nltk.pos_tag(tokens)

print(pos_tags)
````

* ('Pick', 'VB'): 'Pick' is a Verb.
* ('up', 'RP'): 'up' is a particle.
* ('the', 'DT'): 'the' is a determiner.
* ('red', 'JJ'): 'red' is an adjective.
* ('ball', 'NN'): 'ball' is a noun (singular).
* ('near', IN): 'near' is a preposition.

````python
# Example A: Light as a Noun
sent_a = nltk.pos_tag(nltk.word_tokenize(("Turn on the light").lower()))
print(sent_a)

# Example B: Light as an Adjective
sent_b = nltk.pos_tag(nltk.word_tokenize(("This is a light box").lower()))
sent_b
````

### NER

````python
from nltk import word_tokenize, pos_tag, ne_chunk

# Required downloads for NER
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

text = "Go to Lab3 and meet Mary."
tokens = word_tokenize(text)
tags = pos_tag(tokens)

# ne_chunk creates a tree of entities
entities = ne_chunk(tags)
print(entities)
````

#### It identifed Lab3 as a Geo-Political Entity (GPE) and Mary as a PERSON
* Note that Lab3 is not exactly a GPE.
* It is mapped to the closest category to location
* ```NNP``` means Proper Noun
