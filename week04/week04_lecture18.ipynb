
## Problem Statement

Plot the workspace of a simple **2-DOF articulated robot manipulator** using Python.

- Link lengths: $l_1 = 5$, $l_2 = 4$
- Both joints are revolute joints
- The joint angles vary as: $-\pi \le \theta_1 \le \pi$ and $-\pi \le \theta_2 \le \pi$

Generate and plot the reachable workspace of the articulated manipulator in the Cartesian plane.

## Solution

The idea is to sweep both joint angles ($\theta_1$, $\theta_2$) across their full allowed range, compute the resulting end-effector position $(x, y)$ for every combination, and scatter-plot all the points. The cloud of points traced out is the manipulator's reachable workspace.

### Link lengths and joint angle ranges

Define the two link lengths and discretize each joint's angle range into 200 samples between $-\pi$ and $\pi$.

```python
l1 = 5
l2 = 4
theta1 = np.linspace(-np.pi, np.pi, 200)
theta2 = np.linspace(-np.pi, np.pi, 200)
```

### Sweep all joint-angle combinations

For every combination of $\theta_1$ and $\theta_2$, compute the end-effector position using the forward kinematics equations:
$$x = l_1\cos(\theta_1) + l_2\cos(\theta_1 + \theta_2)$$
$$y = l_1\sin(\theta_1) + l_2\sin(\theta_1 + \theta_2)$$

```python
x_points = []
y_points = []
for t1 in theta1:
    for t2 in theta2:
        x = l1 * np.cos(t1) + l2 * np.cos(t1 + t2)
        y = l1 * np.sin(t1) + l2 * np.sin(t1 + t2)
        x_points.append(x)
        y_points.append(y)
```

### Plot the reachable workspace

Scatter-plot every computed $(x, y)$ point to visualize the manipulator's reachable workspace.

```python
plt.figure(figsize=(8,8))
plt.scatter(x_points, y_points, s=1)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Articulated Robot Workspace')
plt.axis('equal')
plt.grid(True)
plt.show()
```
